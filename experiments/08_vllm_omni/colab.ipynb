```python
import os
import time
from pathlib import Path
from urllib.request import urlopen
from urllib.error import URLError

def start(command, name, **options):
    previous = globals().get(name)
    if previous is not None and previous.poll() is None:
        previous.terminate()
        previous.wait(timeout=30)
    log = Path(f"/content/{name}.log")
    with log.open("w") as output:
        return subprocess.Popen(command, stdout=output, stderr=subprocess.STDOUT, **options)

model = "Qwen/Qwen3-TTS-12Hz-0.6B-CustomVoice"
omni = start(["/content/voiceforge-omni/bin/vllm", "serve", model, "--omni",
              "--host", "127.0.0.1", "--port", "8090", "--dtype", "half",
              "--enforce-eager"], "omni")
deadline = time.monotonic() + 600
while time.monotonic() < deadline and omni.poll() is None:
    try:
        with urlopen("http://127.0.0.1:8090/v1/models", timeout=2):
            break
    except (URLError, TimeoutError):
        time.sleep(3)
else:
    print(Path("/content/omni.log").read_text(errors="replace")[-8000:])
    raise RuntimeError("Model server did not become ready. See /content/omni.log.")
gateway = start(["node", "server.js"], "gateway", cwd="/content/voiceforge-smoke/apps/api",
                env={**os.environ, "VLLM_OMNI_URL": "http://127.0.0.1:8090", "PORT": "3001",
                     "VLLM_OMNI_TIMEOUT_MS": "240000"})
time.sleep(2)
if gateway.poll() is not None:
    raise RuntimeError(Path("/content/gateway.log").read_text())
print("Model and gateway started.")
```

```python
import json
import wave
from urllib.request import Request
from urllib.error import URLError
from IPython.display import Audio, display

request = Request("http://127.0.0.1:3001/v1/speech",
                  data=json.dumps({"backend": "vllm", "text": "Hello from VoiceForge.",
                                   "voice_id": "ryan"}).encode(),
                  headers={"Content-Type": "application/json"})
path = Path("/content/voiceforge-vllm.wav")
try:
    with urlopen(request, timeout=255) as response:
        path.write_bytes(response.read())
except (URLError, TimeoutError) as error:
    if hasattr(error, "read"):
        print(error.read().decode(errors="replace"))
    for log in ("gateway", "omni"):
        log_path = Path(f"/content/{log}.log")
        if log_path.exists():
            print(f"--- {log}.log ---")
            print(log_path.read_text(errors="replace")[-8000:])
    raise
with wave.open(str(path)) as wav:
    if not wav.getnframes():
        raise RuntimeError("The WAV contains no audio.")
    print(f"{wav.getframerate()} Hz, {wav.getnframes() / wav.getframerate():.2f} seconds")
display(Audio(filename=str(path)))
```

## Streaming benchmark
Run after the WAV smoke test. The benchmark warms up once, then runs four bursts at each concurrency. With 4–16 requests per level, p95 is the slowest successful request in this smoke benchmark.


```python
import base64
import math
from concurrent.futures import ThreadPoolExecutor
from threading import Barrier
from urllib.error import HTTPError
from google.colab import files

prompt = "Hello from VoiceForge. This is a streaming speech benchmark."
payload = json.dumps({"backend": "vllm", "text": prompt, "voice_id": "ryan"}).encode()

def measure_stream(barrier):
    request = Request("http://127.0.0.1:3001/v1/speech/stream", data=payload,
                      headers={"Content-Type": "application/json"})
    barrier.wait(timeout=30)
    started = time.perf_counter()
    first_ms = None
    rate = None
    pcm_bytes = 0
    chunks = 0
    try:
        with urlopen(request, timeout=255) as response:
            for line in response:
                event = json.loads(line)
                if event["type"] == "start":
                    rate = event["sample_rate"]
                elif event["type"] == "audio":
                    chunk = base64.b64decode(event["pcm"], validate=True)
                    if not chunk or len(chunk) % 2:
                        raise ValueError("Invalid PCM chunk")
                    if first_ms is None:
                        first_ms = (time.perf_counter() - started) * 1000
                    pcm_bytes += len(chunk)
                    chunks += 1
                elif event["type"] == "done" and rate == 24000 and pcm_bytes:
                    return {"first_audio_ms": first_ms,
                            "total_ms": (time.perf_counter() - started) * 1000,
                            "audio_seconds": pcm_bytes / (2 * rate), "chunks": chunks}
                else:
                    raise ValueError(f"Unexpected stream event: {event['type']}")
        raise ValueError("Stream ended without a valid done event")
    except HTTPError as exc:
        return {"status": exc.code, "error": exc.read(4096).decode(errors="replace")}
    except (OSError, ValueError, KeyError) as exc:
        return {"error": str(exc)}

def p95(values):
    return sorted(values)[math.ceil(0.95 * len(values)) - 1] if values else None

warmup = measure_stream(Barrier(1))
if "error" in warmup:
    print(Path("/content/omni.log").read_text(errors="replace")[-8000:])
    raise RuntimeError(warmup)

report = {"backend": "vllm-omni 0.30.0", "model": model, "voice": "ryan",
          "prompt": prompt, "gpu": subprocess.check_output(["nvidia-smi", "-L"], text=True).strip(),
          "rounds_per_concurrency": 4, "results": []}
with ThreadPoolExecutor(max_workers=4) as pool:
    for concurrency in (1, 2, 4):
        bursts = []
        for _ in range(4):
            gate = Barrier(concurrency)
            started = time.perf_counter()
            jobs = [pool.submit(measure_stream, gate) for _ in range(concurrency)]
            rows = [job.result() for job in jobs]
            bursts.append({"wall_seconds": time.perf_counter() - started, "requests": rows})
        completed = [row for burst in bursts for row in burst["requests"] if "error" not in row]
        wall_seconds = sum(burst["wall_seconds"] for burst in bursts)
        result = {"concurrency": concurrency, "completed": len(completed),
                  "failed": concurrency * len(bursts) - len(completed),
                  "p95_first_audio_ms": p95([row["first_audio_ms"] for row in completed]),
                  "p95_total_ms": p95([row["total_ms"] for row in completed]),
                  "audio_seconds_per_second": sum(row["audio_seconds"] for row in completed) / wall_seconds,
                  "requests_per_second": len(completed) / wall_seconds, "bursts": bursts}
        report["results"].append(result)
        print({key: value for key, value in result.items() if key != "bursts"})
        if result["failed"]:
            print(Path("/content/omni.log").read_text(errors="replace")[-8000:])
            break
path = Path("/content/vllm-streaming.json")
path.write_text(json.dumps(report, indent=2))
files.download(str(path))
```
